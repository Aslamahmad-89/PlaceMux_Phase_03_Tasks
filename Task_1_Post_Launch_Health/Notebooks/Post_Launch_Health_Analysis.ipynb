## Stage A — Understand & Set the Bar

### A1. Definition of Done

This task will be considered complete when:

1. A real-data launch performance report is built covering:
   - Funnel
   - Revenue
   - Retention
   - Quality

2. A ranked, evidence-backed list of the biggest problems is created.

3. A Phase-3 analytics backlog is created and tied to decisions.

4. All important metrics are traceable to their source data.

5. The analysis is validated, sanity-checked, and demoable end-to-end.

### A2. Upstream Inputs Available

The following real data from completed Phase-2 work is available:

- Company event data
- Job posting data
- Payment and revenue data
- Marketplace KPI data

No synthetic data will be introduced unless the task explicitly requires it.

### A3. What Does "Good" Mean?

A good launch-health analysis means:

- Decisions are based on real evidence rather than opinion.
- Every important metric is traceable to its source data.
- Funnel, revenue, retention, and quality are evaluated together.
- Problems are ranked by evidence and business impact.
- Data quality and uncertainty are explicitly validated.
- Every identified problem leads to a clear action or decision.
- The complete analysis can be demonstrated end-to-end using real data.

## Stage B — Build a Launch Performance Report

| Area | Metric | Source | Decision |
|---|---|---|---|

No metric will be included only because it is easy to calculate or looks impressive.

```python
# Basic real-data model preparation

# Convert event timestamps to datetime
company_events["event_timestamp"] = pd.to_datetime(
    company_events["event_timestamp"],
    errors="coerce"
)

job_posted_events["event_timestamp"] = pd.to_datetime(
    job_posted_events["event_timestamp"],
    errors="coerce"
)

# Create event-level funnel counts
funnel_counts = (
    company_events["event_name"]
    .value_counts()
    .rename_axis("event_name")
    .reset_index(name="event_count")
)

print("===== LAUNCH FUNNEL =====")
display(funnel_counts)

print("\n===== JOB POSTING SUMMARY =====")
display(job_posted_events.describe(include="all"))

print("\n===== PAYMENT SUMMARY =====")
display(payments_data.describe(include="all"))
```

### B3. Validation, Reconciliation & Uncertainty

The analysis will be validated against the available source datasets.

Validation checks include:

- Row counts and duplicate records
- Missing values
- Invalid or unparseable timestamps
- Validity of event and payment statuses
- Reconciliation of derived metrics with source-level records
- Magnitude and consistency sanity checks

Any limitation caused by unavailable upstream data will be explicitly documented rather than hidden or replaced with synthetic data.

```python
# B3 — Validation, Reconciliation & Uncertainty

datasets = {
    "Company Events": company_events,
    "Job Posted Events": job_posted_events,
    "Payments Data": payments_data
}

# 1. Basic data-quality validation
validation_results = []

for name, df in datasets.items():
    validation_results.append({
        "Dataset": name,
        "Rows": len(df),
        "Columns": len(df.columns),
        "Duplicate Rows": df.duplicated().sum(),
        "Missing Cells": df.isna().sum().sum()
    })

validation_df = pd.DataFrame(validation_results)

print("===== DATA QUALITY VALIDATION =====")
display(validation_df)


# 2. Company event validation
print("===== COMPANY EVENT VALUES =====")
display(company_events["event_name"].value_counts(dropna=False))


# 3. Payment dataset structure
print("===== PAYMENT DATA COLUMNS =====")
print(payments_data.columns.tolist())

print("\n===== PAYMENT DATA SAMPLE =====")
display(payments_data.head())


# 4. Payment categorical columns
print("===== PAYMENT CATEGORICAL VALUES =====")

for col in payments_data.select_dtypes(include=["object"]).columns:
    print(f"\n{col}:")
    display(payments_data[col].value_counts(dropna=False))


# 5. Timestamp validation
print("===== TIMESTAMP VALIDATION =====")

print(
    "Company Events invalid timestamps:",
    company_events["event_timestamp"].isna().sum()
)

print(
    "Job Posted Events invalid timestamps:",
    job_posted_events["event_timestamp"].isna().sum()
)


# 6. Numeric payment columns
print("\n===== PAYMENT NUMERIC COLUMNS =====")

numeric_payment_cols = payments_data.select_dtypes(
    include=["number"]
).columns.tolist()

print(numeric_payment_cols)

if numeric_payment_cols:
    display(payments_data[numeric_payment_cols].describe())
```

## Stage C — Ranked, Evidence-Backed List of the Biggest Problems

### C1. Metric, Source & Decision

The biggest launch problems will be identified using measurable evidence from the available real data.

| Problem Area | Metric / Evidence | Source | Decision |
|---|---|---|---|
| Funnel conversion | Job views vs applications submitted | company_events.csv | Investigate candidate conversion drop-off |
| Job supply | Jobs posted and job-posting activity | company_events.csv / job_posted_events.csv | Identify marketplace supply gaps |
| Revenue reliability | Successful vs failed payments | payments_data.csv | Investigate payment failures and revenue leakage |
| Data quality | Missing values, invalid timestamps, duplicates | All source datasets | Assess whether reported metrics are reliable |
| Retention | Repeat activity after launch | Available event data | Determine whether retention can be measured reliably |

### Problem Ranking Principle

Problems will be ranked using:

1. Magnitude of the observed issue
2. Business impact
3. Reliability of the supporting data
4. Ability to take a concrete action

No problem will be ranked without evidence from the available source data.

```python
# C2 — Evidence-based problem analysis

# 1. Funnel metrics
event_counts = company_events["event_name"].value_counts()

company_signups = event_counts.get("company_signup", 0)
jobs_posted = event_counts.get("job_posted", 0)
job_views = event_counts.get("job_viewed", 0)
applications = event_counts.get("application_submitted", 0)

view_to_application_rate = (
    applications / job_views * 100
    if job_views > 0 else 0
)

signup_to_job_rate = (
    jobs_posted / company_signups * 100
    if company_signups > 0 else 0
)

# 2. Job posting reconciliation
job_event_rows = len(job_posted_events)

# 3. Data-quality evidence
company_missing = company_events.isna().sum().sum()
job_missing = job_posted_events.isna().sum().sum()
payment_missing = payments_data.isna().sum().sum()

# 4. Create evidence table
problem_evidence = pd.DataFrame({
    "Problem Area": [
        "Job-view to application conversion",
        "Company signup to job-posting conversion",
        "Company event data completeness",
        "Job posting data completeness",
        "Payment data completeness"
    ],
    "Evidence": [
        f"{view_to_application_rate:.2f}%",
        f"{signup_to_job_rate:.2f}%",
        f"{company_missing} missing cells",
        f"{job_missing} missing cells",
        f"{payment_missing} missing cells"
    ],
    "Source": [
        "company_events.csv",
        "company_events.csv",
        "company_events.csv",
        "job_posted_events.csv",
        "payments_data.csv"
    ]
})

display(problem_evidence)

print("===== FUNNEL COUNTS =====")
print(f"Company Signups: {company_signups}")
print(f"Jobs Posted: {jobs_posted}")
print(f"Job Views: {job_views}")
print(f"Applications: {applications}")

print("\n===== CONVERSION METRICS =====")
print(f"View → Application: {view_to_application_rate:.2f}%")
print(f"Signup → Job Posted: {signup_to_job_rate:.2f}%")

print("\n===== JOB POSTING RECONCILIATION =====")
print(f"Rows in job_posted_events.csv: {job_event_rows}")
print(f"job_posted events in company_events.csv: {jobs_posted}")
```

```python
# C3 — Validate the evidence behind identified problems

print("===== C3: PROBLEM VALIDATION =====")

# 1. Funnel reconciliation
company_event_total = len(company_events)
calculated_funnel_total = (
    company_signups +
    jobs_posted +
    job_views +
    applications
)

print("\nFunnel event reconciliation:")
print("Total company event rows:", company_event_total)
print("Sum of selected funnel events:", calculated_funnel_total)

# 2. Job-posting reconciliation
print("\nJob posting reconciliation:")
print("job_posted events in company_events:", jobs_posted)
print("Rows in job_posted_events:", job_event_rows)

# 3. Conversion sanity checks
print("\nConversion sanity checks:")
print(f"View → Application rate: {view_to_application_rate:.2f}%")
print(f"Signup → Job Posted rate: {signup_to_job_rate:.2f}%")

# 4. Data-quality validation
print("\nData-quality checks:")
print("Company events duplicates:", company_events.duplicated().sum())
print("Job posting duplicates:", job_posted_events.duplicated().sum())
print("Payment duplicates:", payments_data.duplicated().sum())

print("Company events missing cells:", company_missing)
print("Job posting missing cells:", job_missing)
print("Payment missing cells:", payment_missing)

# 5. Retention limitation
print("\nRetention validation:")
print(
    "Retention cannot currently be confirmed from the available "
    "Phase-2 datasets because a reliable user-level longitudinal "
    "activity source has not yet been established."
)
```

### C4. Rank Problems by Evidence and Business Impact

The identified problems are ranked using:

1. Magnitude of the observed issue
2. Business impact
3. Reliability of the supporting data
4. Ability to take a concrete action

Each ranked problem is linked to evidence, source data, and a recommended action.

```python
# C4 — Ranked Evidence-Backed Problem List

ranked_problems = pd.DataFrame({
    "Rank": [1, 2, 3, 4],
    "Problem": [
        "Job-view to application conversion",
        "Company signup to job-posting conversion",
        "Payment reliability",
        "Data completeness / source coverage"
    ],
    "Evidence": [
        f"View → Application conversion = {view_to_application_rate:.2f}%",
        f"Signup → Job Posted conversion = {signup_to_job_rate:.2f}%",
        "Payment success/failure performance requires payment-status validation",
        f"Company events contain {company_missing} missing cells"
    ],
    "Source": [
        "company_events.csv",
        "company_events.csv",
        "payments_data.csv",
        "Phase-2 source datasets"
    ],
    "Business Impact": [
        "High",
        "High",
        "High",
        "Medium"
    ],
    "Recommended Action": [
        "Investigate candidate funnel drop-off and improve job-to-application conversion",
        "Investigate why companies do not progress from signup to job posting",
        "Identify payment failure patterns and reduce revenue leakage",
        "Perform field-level missingness and source-coverage analysis before using affected metrics"
    ]
})

display(ranked_problems)
```
